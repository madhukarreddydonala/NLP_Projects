# spam classification 

```python
# nlp practice 
# spam classification 
import pandas as pd 
mesge=pd.read_csv('SMSSpamCollection.txt',sep='\t',names=['label','message'])

```

```python
mesge
```

```python
# data cleaning and preprocessing 
import re 
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
ps=PorterStemmer()
lemmatizer=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(0,len(mesge)):
    review=re.sub('[^a-zA-Z]',' ',mesge['message'][i])
    review=review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
len(corpus)
```

```python
# out put features 
y=pd.get_dummies(mesge['label'],dtype='float')
y
```

```python
y=y.iloc[:,0].values
```

```python
#train test split 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
X_train
```

```python
len(X_train),len(y_train)
```

```python
# create BOW 
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500,ngram_range=(1,2))
```

```python
X_train=cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
X_train
```

```python
from sklearn.naive_bayes import MultinomialNB
spmamodel=MultinomialNB().fit(X_train,y_train)
```

```python
spmamodel
```

```python
y_pred=spmamodel.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
classification_report(y_test,y_pred)
confusion_m=confusion_matrix(y_test,y_pred)
accuracy=accuracy_score(y_test,y_pred)
print(confusion_m)
print(accuracy)
print(classification_report(y_test,y_pred))
```

```python
# test with new messages
#from textpreprocessing import preprocess_text
new_messages = ["Congratulations! You've won a free iPhone!", "Hey, are we still on for lunch tomorrow?"]
new_corpus=[]
for i in range(0,len(new_messages)):
    review=re.sub('[^a-zA-Z]',' ',new_messages[i])
    review=review.lower()
    review=review.split()
    review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    new_corpus.append(review)
    
```

```python
#test cleaning and preprocessing
# convert to vector using the same CountVectorizer
new_vectors = cv.transform(new_corpus).toarray()
new_vectors
y_pred = spmamodel.predict(new_vectors)
y_pred
# compare the predictions with the actual labels
```

# TF-IDF

```python
#train test split 
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.20)
```

```python
X_train
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=2500, ngram_range=(1,3))
X_train = tfidf.fit_transform(X_train)
X_test = tfidf.transform(X_test)
```

```python
tfidf.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
spmamodel=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred=spmamodel.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
classification_report(y_test,y_pred)
confusion_m=confusion_matrix(y_test,y_pred)
accuracy=accuracy_score(y_test,y_pred)
print(confusion_m)
print(accuracy)
print(classification_report(y_test,y_pred))
```
